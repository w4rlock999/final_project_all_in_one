```python
# To handle keys that are still JSON strings, we can recursively parse any value that looks like a JSON string.
def try_json_load(val):
    if isinstance(val, str):
        try:
            loaded = json.loads(val)
            # If loaded is a dict or list, recursively process it
            if isinstance(loaded, dict):
                return {k: try_json_load(v) for k, v in loaded.items()}
            elif isinstance(loaded, list):
                return [try_json_load(x) for x in loaded]
            else:
                return loaded
        except Exception:
            return val
    elif isinstance(val, dict):
        return {k: try_json_load(v) for k, v in val.items()}
    elif isinstance(val, list):
        return [try_json_load(x) for x in val]
    else:
        return val

trace_dict = json.loads(traces[0].to_json())
trace_dict = try_json_load(trace_dict)

with open("./data/input/langgraph_trace_example_x.json", "w") as f:
    json.dump(trace_dict, f, indent=2, default=str)
```

# Trace To Detailed Graph


```python
import json

with open('./data/input/langgraph_trace_example.json', 'r') as f:
    langgraph_trace = json.load(f)

print(len(langgraph_trace['data']['spans']))

import os
from trace_to_graph_langgraph import TraceGraphFromLangGraph
import json 

def run_trace_to_graph(trace_file: str):
 
    # Create TraceGraph instance and process the trace
    # Generate the detailed graph with relationships
    detailed_graph = TraceGraphFromLangGraph(trace_file).generate_detailed_knowledge_graph()
    
    current_dir = os.getcwd()

    # Save the results
    output_dir = os.path.join(current_dir, 'data', 'output')
    os.makedirs(output_dir, exist_ok=True)
    
    # Save detailed graph
    with open(os.path.join(output_dir, 'detailed_graph_langgraph_new_format.json'), 'w') as f:
        json.dump(detailed_graph, f, indent=2)

run_trace_to_graph("./data/input/langgraph_trace_example.json")
```
